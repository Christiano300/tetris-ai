```python
import torch
from torch import nn
from tensordict.nn.distributions import NormalParamExtractor
from torchrl.modules import MLP
```

```python
module = nn.Linear(3, 5)
normal_params = NormalParamExtractor()
tensor = torch.randn(3)
out = module(tensor)
print(out)
print(normal_params(out))
```

```python
mlp = MLP(in_features=3, out_features=5, num_cells=[4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4, 4])
out2 = mlp(tensor)
print(out2)
```
